```python
import numpy as np
from functools import lru_cache
```

```python
target = 100
winning_prob = 0.5
@lru_cache
def get_optimal_policy(budget, remaining_round):
    global target, winning_prob
    if remaining_round == 0:
        if budget == target:
            return 1., -1
        else:
            return 0., -1
    if budget > 100 or budget < 0:
        return  0., -1
    cur_optimal_winning_prob = 0
    best_bet = 0
    for bet in range(budget + 1):
        cur_winning_prob = winning_prob * get_optimal_policy(budget + bet, remaining_round - 1)[0] + \
        (1 - winning_prob) * get_optimal_policy(budget - bet, remaining_round - 1)[0]
        if cur_optimal_winning_prob < cur_winning_prob:
            cur_optimal_winning_prob = cur_winning_prob
            best_bet = bet
    return cur_optimal_winning_prob, best_bet
```

```python
bet_policy = []
budget_path = [[75]]
global_optimal_winning_prob = 0

n_round = 3

for tmp_round in range(n_round, 0, -1):
    bet_policy.append([])
    budget_path.append([])
    for tmp_budget in budget_path[-2]:
        optimal_winning_prob, optimal_bet = get_optimal_policy(budget=tmp_budget, remaining_round=tmp_round)
        if tmp_round == n_round:
            global_optimal_winning_prob = optimal_winning_prob
        bet_policy[-1].append(optimal_bet)
        if optimal_bet == 0:
            budget_path[-1].append(cur_budget)#There is only 1 possible budget outcome
        else:#There are 2 cases: win => new_budget = budget + bet and  new_budget = budget - bet
            budget_path[-1].append(cur_budget + optimal_bet)
            budget_path[-1].append(cur_budget - optimal_bet)
print(bet_policy)
```
