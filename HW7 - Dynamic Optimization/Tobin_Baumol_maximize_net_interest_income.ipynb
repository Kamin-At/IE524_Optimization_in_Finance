```python
r = 0.01
E = 0.5
F = 10000
c = 1
p = 100
b = 6000000
```

```python
def v(b):
    global r, x, E, F, c, p
    return r * x * (b - x) / E - 2 * (F + x*c/p)
```

```python
best_x = 0
best_val = 0
for i in range(0, b + 100, 100):
    x = i
    tmp_val = v(b)
    if tmp_val > best_val:
        best_x = x
        best_val = tmp_val
```
